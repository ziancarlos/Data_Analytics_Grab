```python
import pandas as pd
```

```python
df_merchant = pd.read_csv("../data/cleaned/grab_merchants_cleaned.csv")
df_food_order = pd.read_csv("../data/cleaned/grab_food_orders_cleaned.csv")
df_user = pd.read_csv("../data/cleaned/users_cleaned.csv")
df_merchant.head()
```

```python
df_merchant["city"].unique()
```

```python
city_coordinates = {
    "Jakarta": (-6.2088, 106.8456),
    "Bandung": (-6.9175, 107.6191),
    "Surabaya": (-7.2575, 112.7521),
    "Bekasi": (-6.2383, 106.9756),
    "Depok": (-6.4025, 106.7942),
    "Bogor": (-6.5971, 106.8060),
    "Tangerang": (-6.1783, 106.6319),
    "Medan": (3.5952, 98.6722),
    "Makassar": (-5.1477, 119.4327),
    "Bali": (-8.3405, 115.0920), 
}


df_merchant["latitude"] = df_merchant["city"].apply(lambda x: city_coordinates[x][0])
df_merchant["longitude"] = df_merchant["city"].apply(lambda x: city_coordinates[x][1])

```

```python
df_merchant.to_csv("../data/transformed/grab_merchant_transformed.csv", index=False)
df_food_order.to_csv("../data/transformed/grab_food_order_transformed.csv", index=False)
df_user.to_csv("../data/transformed/grab_user_transformed.csv", index=False)
```
